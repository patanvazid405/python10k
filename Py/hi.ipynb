```python
class Movie:
    def __init__(self):
        print("Welcome to INOX")

    def movies(self):
        self.m1 = "coolie"   
        self.m2 = "War2"   
        self.m3 = "Kingdom" 
        print("---Showing Movies---")
        print(f"1.{self.m1} 2.{self.m2} 3.{self.m3}") 

class user(Movie):
    def __init__(self):
        self.inp = input("Enter a movie:")
        if self.inp == super().m1:
            print(super().m1)
        elif self.inp == super().m2:
            print(super().m2)
        elif self.inp == super().m3:
            print(super().m3)        

```
